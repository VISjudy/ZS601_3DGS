```python
# Official 2DGS DTU scan105 validation, then ZS601 B (LiDAR init)
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
import tarfile, subprocess, sys, os, json, time
os.environ['TORCH_CUDA_ARCH_LIST']='8.9'
subprocess.run(['nvidia-smi'],check=True)
archive=Path('/content/drive/MyDrive/LCCDataset/zs601_output/parallel_runtime_assets/zs601_2dgs_patched_runtime.tar.gz')
assert archive.exists(), archive
root=Path('/content/2d-gaussian-splattingWithMask')
if not root.exists():
    with tarfile.open(archive) as tf: tf.extractall('/content')
launcher=root/'colab/run_parallel_experiment.py'
subprocess.run([sys.executable,'-m','py_compile',str(launcher)],check=True)
print('SERIAL PLAN: official DTU scan105 -> ZS601 B',flush=True)
print('LAUNCHER:',launcher,flush=True)
official_rc=subprocess.call([sys.executable,'-u',str(launcher),'--mode','official'])
print('OFFICIAL_RETURN_CODE',official_rc,flush=True)
if official_rc != 0:
    raise RuntimeError('Official DTU validation failed; B is intentionally paused.')
b_rc=subprocess.call([sys.executable,'-u',str(launcher),'--mode','b'])
print('B_RETURN_CODE',b_rc,flush=True)
if b_rc != 0:
    raise RuntimeError('ZS601 B failed; inspect Drive status and logs.')
print('OFFICIAL AND B COMPLETE',flush=True)
```
